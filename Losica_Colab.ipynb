# Build Losica

Choose **Runtime → Run all**. The notebook downloads the public repository, runs the causal acceptance tests, generates a large language, validates it, and downloads `losica-generated-language.zip`.

```
!rm -rf /content/Losica
!git clone --depth 1 https://github.com/silvaguilhermejaime-spec/Losica.git /content/Losica
%cd /content/Losica
!python -m pip install -q -e '.[test]'
!python -m pytest -q tests/test_causal_engine.py
!mkdir -p dist
!python FINALIZE_LANGUAGE.py --seed 19020 --vocabulary-scale large --out dist/language.json | tee dist/build-report.json
!python VALIDATE_COMPLETE_LANGUAGE.py dist/language.json | tee dist/validation-report.json
```

```
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile
from google.colab import files

target = Path('/content/losica-generated-language.zip')
with ZipFile(target, 'w', ZIP_DEFLATED, compresslevel=9) as archive:
    for path in [Path('dist/language.json'), Path('dist/build-report.json'), Path('dist/validation-report.json'), Path('README.md'), Path('CAUSAL_ARCHITECTURE.md'), Path('METHODOLOGY.md'), Path('inputs.lock.json')]:
        archive.write(path, path.name)
files.download(str(target))
```
